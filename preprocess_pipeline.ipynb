```python
def design_bandpass_filter(
    lowcut,
    highcut,
    fs,
    order=4
):
    
    nyquist = 0.5 * fs
    
    low = lowcut / nyquist
    high = highcut / nyquist

    b, a = butter(
        order,
        [low, high],
        btype="band"
    )

    return b, a


def bandpass_filter(
    x,
    lowcut=0.5,
    highcut=45.0,
    fs=256,
    order=4
):
    
    b, a = design_bandpass_filter(
        lowcut,
        highcut,
        fs,
        order
    )

    return filtfilt(
        b,
        a,
        x
    )
```

```python
filtered_signal = bandpass_filter(
    eeg_signal,
    lowcut=LOWCUT,
    highcut=HIGHCUT,
    fs=FS,
    order=FILTER_ORDER
)

print("Bandpass filtering completed.")
```

```python
def calculate_band_power(
    x,
    fs,
    bands
):

    f, psd = welch(
        x,
        fs=fs,
        nperseg=min(2048, len(x))
    )

    results = {}

    for band, (low, high) in bands.items():

        mask = (
            (f >= low) &
            (f <= high)
        )

        if np.any(mask):

            power = np.trapezoid(
                psd[mask],
                f[mask]
            )

        else:

            power = 0.0

        results[band] = power

    return results
```

```python
# ============================================================
# IMPROVED EEG WAVELET SCALOGRAM
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
import pywt

# ------------------------------------------------------------
# Parameters
# ------------------------------------------------------------

FS = 256

WAVELET = "morl"

FREQ_MIN = 0.5
FREQ_MAX = 45

# Analyze first N seconds
# Use the complete signal if you want the full recording
DISPLAY_SECONDS = 30

N_SAMPLES = min(
    int(DISPLAY_SECONDS * FS),
    len(filtered_signal)
)

wavelet_signal = filtered_signal[:N_SAMPLES]

# ------------------------------------------------------------
# Determine scales from desired frequencies
# ------------------------------------------------------------

central_frequency = pywt.central_frequency(
    WAVELET
)

frequencies = np.geomspace(
    FREQ_MIN,
    FREQ_MAX,
    128
)

scales = (
    central_frequency * FS
) / frequencies

# ------------------------------------------------------------
# Continuous Wavelet Transform
# ------------------------------------------------------------

coefficients, frequencies_cwt = pywt.cwt(
    wavelet_signal,
    scales,
    WAVELET,
    sampling_period=1 / FS
)

# ------------------------------------------------------------
# Sort frequencies from low → high
# ------------------------------------------------------------

sort_idx = np.argsort(
    frequencies_cwt
)

frequencies_cwt = frequencies_cwt[
    sort_idx
]

coefficients = coefficients[
    sort_idx,
    :
]

# ------------------------------------------------------------
# Wavelet power
# ------------------------------------------------------------

wavelet_power = np.abs(
    coefficients
) ** 2

# Convert to dB
wavelet_power_db = 10 * np.log10(
    wavelet_power + 1e-12
)

# Normalize
wavelet_power_db = (
    wavelet_power_db
    - np.max(wavelet_power_db)
)

# ------------------------------------------------------------
# Time axis
# ------------------------------------------------------------

time_wavelet = (
    np.arange(N_SAMPLES) / FS
)

# ------------------------------------------------------------
# Plot
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(16, 7)
)

mesh = ax.pcolormesh(
    time_wavelet,
    frequencies_cwt,
    wavelet_power_db,
    shading="auto",
    cmap="turbo",
    vmin=-50,
    vmax=0
)

# ------------------------------------------------------------
# EEG frequency bands
# ------------------------------------------------------------

band_limits = [
    0.5,
    4,
    8,
    13,
    30,
    45
]

for freq in band_limits:

    ax.axhline(
        freq,
        linestyle="--",
        linewidth=0.8,
        alpha=0.6
    )

# ------------------------------------------------------------
# Band labels
# ------------------------------------------------------------

band_labels = [
    ("Delta", 2.0),
    ("Theta", 6.0),
    ("Alpha", 10.5),
    ("Beta", 21.5),
    ("Gamma", 37.5)
]

for label, freq in band_labels:

    ax.text(
        time_wavelet[-1] * 1.005,
        freq,
        label,
        va="center",
        fontsize=10
    )

# ------------------------------------------------------------
# Axis
# ------------------------------------------------------------

ax.set_ylim(
    FREQ_MIN,
    FREQ_MAX
)

ax.set_title(
    "EEG Continuous Wavelet Transform (Scalogram)",
    fontsize=16,
    fontweight="bold"
)

ax.set_xlabel(
    "Time (seconds)",
    fontsize=12
)

ax.set_ylabel(
    "Frequency (Hz)",
    fontsize=12
)

# ------------------------------------------------------------
# Colorbar
# ------------------------------------------------------------

cbar = fig.colorbar(
    mesh,
    ax=ax,
    pad=0.02
)

cbar.set_label(
    "Wavelet power relative to maximum (dB)",
    fontsize=11
)

# ------------------------------------------------------------
# Grid
# ------------------------------------------------------------

ax.grid(
    True,
    linestyle=":",
    alpha=0.25
)

plt.tight_layout()
plt.show()
```

```python
def segment_signal(
    x,
    window_samples,
    step_samples
):

    windows = []

    starts = range(
        0,
        len(x) - window_samples + 1,
        step_samples
    )

    for start in starts:

        end = start + window_samples

        windows.append(
            x[start:end]
        )

    return np.asarray(
        windows,
        dtype=np.float32
    )
```

```python
def calculate_spectral_entropy(
    x,
    fs
):

    f, psd = welch(
        x,
        fs=fs,
        nperseg=min(2048, len(x))
    )

    psd = np.maximum(
        psd,
        1e-12
    )

    probability = (
        psd /
        np.sum(psd)
    )

    entropy = -np.sum(
        probability *
        np.log2(probability)
    )

    return entropy


entropy = calculate_spectral_entropy(
    filtered_signal,
    FS
)

print(
    f"Spectral entropy: {entropy:.5f}"
)
```

```python
def hjorth_parameters(x):

    activity = np.var(x)

    dx = np.diff(x)
    ddx = np.diff(dx)

    mobility = np.sqrt(
        np.var(dx) /
        (activity + 1e-12)
    )

    mobility_derivative = np.sqrt(
        np.var(ddx) /
        (np.var(dx) + 1e-12)
    )

    complexity = (
        mobility_derivative /
        (mobility + 1e-12)
    )

    return (
        activity,
        mobility,
        complexity
    )


activity, mobility, complexity = hjorth_parameters(
    filtered_signal
)

print("=" * 60)
print("HJORTH PARAMETERS")
print("=" * 60)

print(f"Activity   : {activity:.6f}")
print(f"Mobility   : {mobility:.6f}")
print(f"Complexity : {complexity:.6f}")
```

```python
from scipy.stats import skew, kurtosis

def extract_features_from_window(
    x,
    fs=256
):

    features = {}

    # Time-domain
    features["mean"] = np.mean(x)
    features["std"] = np.std(x)
    features["variance"] = np.var(x)
    features["rms"] = np.sqrt(np.mean(x ** 2))
    features["min"] = np.min(x)
    features["max"] = np.max(x)
    features["peak_to_peak"] = np.ptp(x)

    features["skewness"] = skew(x)
    features["kurtosis"] = kurtosis(x)

    # Zero crossing rate
    zero_crossings = np.sum(
        np.diff(np.signbit(x))
    )

    features["zero_crossing_rate"] = (
        zero_crossings / len(x)
    )

    # Hjorth
    activity, mobility, complexity = (
        hjorth_parameters(x)
    )

    features["hjorth_activity"] = activity
    features["hjorth_mobility"] = mobility
    features["hjorth_complexity"] = complexity

    # Spectral features
    f, psd = welch(
        x,
        fs=fs,
        nperseg=min(1024, len(x))
    )

    total_power = np.trapezoid(
        psd,
        f
    )

    features["total_power"] = total_power

    # Band powers
    for band, (low, high) in EEG_BANDS.items():

        mask = (
            (f >= low) &
            (f <= high)
        )

        power = np.trapezoid(
            psd[mask],
            f[mask]
        )

        features[
            f"{band.lower()}_power"
        ] = power

        features[
            f"{band.lower()}_relative_power"
        ] = power / (
            total_power + 1e-12
        )

    # Spectral entropy
    features[
        "spectral_entropy"
    ] = calculate_spectral_entropy(
        x,
        fs
    )

    return features
```

```python
def preprocess_dataframe(
    dataframe,
    signal_columns,
    fs=256,
    lowcut=0.5,
    highcut=45.0,
    filter_order=4,
    window_seconds=4,
    overlap=0.5
):

    window_samples = int(
        fs * window_seconds
    )

    step_samples = int(
        window_samples * (1 - overlap)
    )

    X_windows = []
    y_windows = []

    for _, row in tqdm(
        dataframe.iterrows(),
        total=len(dataframe),
        desc="Processing recordings"
    ):

        x = row[
            signal_columns
        ].values.astype(np.float32)

        # Filtering
        x_filtered = bandpass_filter(
            x,
            lowcut,
            highcut,
            fs,
            filter_order
        )

        # Normalization
        x_filtered = zscore_signal(
            x_filtered
        )

        # Windowing
        windows = segment_signal(
            x_filtered,
            window_samples,
            step_samples
        )

        # Store windows
        X_windows.append(
            windows
        )

        # Same label for all windows from this source row
        y_windows.extend(
            [row[TARGET_COL]] *
            len(windows)
        )

    if len(X_windows) == 0:

        return (
            np.empty(
                (0, window_samples),
                dtype=np.float32
            ),
            np.empty(
                (0,),
                dtype=np.int64
            )
        )

    X_windows = np.concatenate(
        X_windows,
        axis=0
    )

    y_windows = np.asarray(
        y_windows
    )

    return (
        X_windows,
        y_windows
    )
```
